```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt

# probabilités
knn_prob = knn.predict_proba(X_test)[:,1]
tree_prob = tree.predict_proba(X_test)[:,1]
rf_prob = rf.predict_proba(X_test)[:,1]

# ROC KNN
fpr1, tpr1, _ = roc_curve(y_test, knn_prob)
roc1 = auc(fpr1, tpr1)

# ROC Tree
fpr2, tpr2, _ = roc_curve(y_test, tree_prob)
roc2 = auc(fpr2, tpr2)

# ROC RF
fpr3, tpr3, _ = roc_curve(y_test, rf_prob)
roc3 = auc(fpr3, tpr3)

plt.figure(figsize=(7,6))

plt.plot(fpr1, tpr1, label=f"KNN (AUC={roc1:.2f})")
plt.plot(fpr2, tpr2, label=f"Decision Tree (AUC={roc2:.2f})")
plt.plot(fpr3, tpr3, label=f"Random Forest (AUC={roc3:.2f})")

plt.plot([0,1],[0,1],"--")

plt.title("Comparaison des modèles - ROC Curve")
plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.legend()
plt.show()
```
